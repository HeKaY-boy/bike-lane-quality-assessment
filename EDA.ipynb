```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_candidates(file_list, title_prefix="Candidate"):
    fig, axes = plt.subplots(nrows=len(file_list), ncols=1, figsize=(12, 4 * len(file_list)), sharex=False, sharey=True)
    
    # Handle single file edge case
    if len(file_list) == 1:
        axes = [axes]
        
    for ax, file in zip(axes, file_list):
        df = pd.read_csv(file)
        
        # Plot x, y, z axes
        ax.plot(df['time'], df['x'], label='x-axis', alpha=0.7)
        ax.plot(df['time'], df['y'], label='y-axis', alpha=0.7)
        ax.plot(df['time'], df['z'], label='z-axis', alpha=0.7)
        
        ax.set_title(f"{title_prefix}: {file}")
        ax.set_ylabel("Acceleration (m/s²)")
        ax.legend(loc='upper right')
        ax.grid(True, linestyle='--', alpha=0.5)

    plt.xlabel("Time")
    plt.tight_layout()
    plt.show()

# Example usage for your smooth candidate files
himanshu_smooth_files = ['data_raw/smooth/Himanshu/Smooth_21/Accelerometer.csv', 'data_raw/smooth/Himanshu/Smooth_21 (2)/Accelerometer.csv', 'data_raw/smooth/Himanshu/Smooth-2026-09-13_14-36-00/Accelerometer.csv']
plot_candidates(himanshu_smooth_files, "Smooth Road Comparison")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_additional_sensors(file_list, sensor_name, y_label):
    fig, axes = plt.subplots(nrows=len(file_list), ncols=1, figsize=(12, 4 * len(file_list)), sharex=False, sharey=True)
    
    if len(file_list) == 1:
        axes = [axes]
        
    for ax, file in zip(axes, file_list):
        df = pd.read_csv(file)
        
        # Convert epoch nanoseconds to relative seconds for a cleaner x-axis
        time_col = 'time' if 'time' in df.columns else 'Time'
        df['relative_time'] = (df[time_col] - df[time_col].min()) / 1e9
        
        ax.plot(df['relative_time'], df['x'], label='x-axis', alpha=0.7)
        ax.plot(df['relative_time'], df['y'], label='y-axis', alpha=0.7)
        ax.plot(df['relative_time'], df['z'], label='z-axis', alpha=0.7)
        
        ax.set_title(f"{sensor_name} Candidate: {file.split('/')[-2]}/{file.split('/')[-1]}")
        ax.set_ylabel(y_label)
        ax.legend(loc='upper right')
        ax.grid(True, linestyle='--', alpha=0.5)

    plt.xlabel("Time (seconds)")
    plt.tight_layout()
    plt.show()

# Example usage for Gyroscope
gyro_files = [
    'data_raw/smooth/Himanshu/Smooth_21/Gyroscope.csv', 
    'data_raw/smooth/Himanshu/Smooth_21 (2)/Gyroscope.csv',
    'data_raw/smooth/Himanshu/Smooth-2026-09-13_14-36-00/Gyroscope.csv'
]
plot_additional_sensors(gyro_files, "Gyroscope", "Angular Velocity (rad/s)")

# Example usage for Gravity
gravity_files = [
    'data_raw/smooth/Himanshu/Smooth_21/Gravity.csv', 
    'data_raw/smooth/Himanshu/Smooth_21 (2)/Gravity.csv',
    'data_raw/smooth/Himanshu/Smooth-2026-09-13_14-36-00/Gravity.csv'
]
plot_additional_sensors(gravity_files, "Gravity", "Acceleration (m/s²)")
```
